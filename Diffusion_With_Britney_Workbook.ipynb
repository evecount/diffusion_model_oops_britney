## 💡 **Core Thesis: Why Character Consistency Is Table Stakes—And What AI Video Is Actually Missing**

Standard lectures on **Types of Generation Models** split the world into three neat boxes:
1. **Autoregressive Models (AR):** *"Looks a little bit behind, and then steps forward."* ($P(x_t \mid x_{<t})$)
2. **Encoder-Decoder Models:** *"Looks both ways—wants to know everything in the input before emitting a high-confidence output."*
3. **Diffusion Models:** *"Creative propagation into underspecified space."* When a human asks for *"a horse in a grassy background"*, the human hasn't specified the exact horse color or whether the grass is a forest, lawn, paddock, or field—and none of those diffusion samples are wrong!

### **The Next Frontier: Autoregressive-Guided Diffusion + Director of Photography (DP) Transitions**
- **The Church Door & Brass Handle Invariant:** If Shot 1 shows *a door in front of a church*, and Shot 2 shows *the door opening*, Shot 2 **must** preserve the exact same door color, the exact same handle geometry, and know whether a person is standing behind the door opening it. An **Autoregressive World-State Model running in the background outside the human's visible prompt** must lock and feed those invariants into the Diffusion denoiser.
- **Why Music Videos Are the Ultimate SFT Benchmark:** Today's generative AI companies can generate consistent character faces, but they lack the craft of a **Director of Photography (DP) and Music Video Editor**, where **the 0.35s transition is as evocative as the main scene**, and **3 disparate visual movements** intercut seamlessly because they are locked to the **same underlying music track**.
